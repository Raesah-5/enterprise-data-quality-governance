# Stage 03 — Data Quality Issues & Findings

## Enterprise Data Quality & Governance Project

### Objective
This stage interprets the results of the Data Quality Assessment and separates:
- Data Quality violations
- Standardization observations
- Resolved findings
- Assessment limitations

The goal is not only to identify problems, but also to understand what the results actually mean from a Data Quality and Governance perspective.


## 1. Load the Dataset

The dataset is loaded again so that the observations identified during profiling can be examined independently.


```python
import pandas as pd

df = pd.read_csv("customers.csv")
df["signup_date"] = pd.to_datetime(df["signup_date"], errors="coerce")

df.head()
```

## 2. Review the Identified Observations

During Stage 01, two notable standardization observations were identified:
- `LOW` in `income_level`
- `Chennal` in `location`

These values are important because they can affect consistency, filtering, grouping, reporting, and downstream data use.


### 2.1 Income Level Observation

```python
df["income_level"].value_counts()
```

### Finding

The dataset contains `LOW` while the expected standardized representation is `Low`.

This is primarily a **consistency / standardization issue**. It does not automatically prove that the underlying customer's income classification is inaccurate.


### 2.2 Location Observation

```python
df["location"].value_counts()
```

### Finding

The dataset contains `Chennal`, which appears to be a misspelled representation of `Chennai`.

This is treated as a **standardization observation**. Without an external source of truth, it is safer to describe the issue as a data representation problem rather than claim that the customer's actual location is inaccurate.


## 3. Standardize the Identified Values

The identified representations are standardized so that equivalent values use one consistent form.


```python
df["income_level"] = df["income_level"].replace({
    "LOW": "Low"
})

df["location"] = df["location"].replace({
    "Chennal": "Chennai"
})

df["income_level"].value_counts(), df["location"].value_counts()
```

### Standardization Result

After standardization:
- `LOW` → `Low`
- `Chennal` → `Chennai`

The purpose of this step is to establish a consistent representation before the final validation state.


## 4. Verify the Standardization

The following checks confirm that the inconsistent representations are no longer present.


```python
remaining_income_issue = (df["income_level"] == "LOW").sum()
remaining_location_issue = (df["location"] == "Chennal").sum()

print("Remaining LOW values:", remaining_income_issue)
print("Remaining Chennal values:", remaining_location_issue)
```

## 5. Formal Data Quality Result

The formal assessment from Stage 02 resulted in:

| Metric | Result |
|---|---:|
| Data Quality Rules | 15 |
| Violations | 0 |
| Quality Rate | 100% |
| Overall Status | PASS |

The important distinction is that the profiling observations were investigated and standardized, while the final formal rule assessment contained no remaining violations.


## 6. Findings vs. Violations

A key lesson from this stage is that not every observation should automatically be treated as a confirmed data quality failure.

### Standardization observations
- `LOW` → `Low`
- `Chennal` → `Chennai`

These were resolved as consistency/standardization issues.

### Formal validation result
- 15 rules assessed
- 0 violations
- 100% quality rate
- PASS

Therefore, the absence of formal violations is a valid result. A Data Quality assessment does not need to produce a large number of errors to be useful.


## 7. Accuracy Limitation

Accuracy was not formally assessed.

To determine whether a value is actually accurate, a trusted external source of truth would be required.

For example, the dataset alone cannot prove whether:
- a customer's recorded location is their real location
- a customer's income level is factually correct

Therefore, this project documents Accuracy as an **assessment limitation**, not as a failed Data Quality dimension.


## 8. Timeliness Limitation

Timeliness was not formally assessed.

The dataset does not provide:
- a `last_updated_date`
- a data refresh timestamp
- a defined business expectation for how frequently the data should be updated

Without these requirements, a Timeliness score would not be evidence-based.


## 9. Key Learning

At the beginning of the project, it was expected that the dataset would contain many obvious Data Quality problems.

The assessment showed something more important: **a good Data Quality process is not measured by the number of problems discovered.**

The value comes from:
1. Defining clear quality rules.
2. Measuring the data against those rules.
3. Separating observations from confirmed violations.
4. Documenting what can and cannot be assessed.
5. Providing evidence for the final result.

A result of **0 violations and 100% quality rate** can therefore be a meaningful Data Quality finding when the rules and limitations are clearly documented.


# Stage 03 Conclusion

The analysis identified two standardization observations: `LOW` in `income_level` and `Chennal` in `location`. Both were standardized to their expected representations.

The formal Data Quality assessment remained at **15 rules, 0 violations, 100% quality rate, and PASS**.

The stage also highlighted two important assessment limitations: Accuracy could not be proven without an external source of truth, and Timeliness could not be evaluated without update metadata or business refresh requirements.

These findings provide the foundation for **Stage 04 — Data Quality Validation Framework**, where the rules and validation logic are organized into a more reusable framework.

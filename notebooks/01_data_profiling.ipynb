# Stage 01 — Data Profiling

## Enterprise Data Quality & Governance Project

### Objective
The purpose of this stage is to understand the structure and basic characteristics of the Customer 360 dataset before performing formal Data Quality Assessment.

The profiling focuses on:
- Dataset dimensions and structure
- Data types
- Missing values
- Duplicate records
- Uniqueness of `customer_id`
- Basic distributions and value ranges
- Date validity and range
- Initial observations that may require further quality assessment


## 1. Load the Dataset

The dataset used in this project is `customers.csv`.

```python
import pandas as pd

df = pd.read_csv("customers.csv")
df.head()
```

## 2. Dataset Dimensions

First, we check the number of records and columns.

```python
df.shape
```

**Observed result:** 100,000 records and 6 columns.

## 3. Dataset Structure and Data Types

The `info()` method provides an overview of column names, data types, and non-null values.

```python
df.info()
```

### Initial observations
- `customer_id` is an integer field.
- `age` is an integer field.
- `gender`, `location`, and `income_level` are categorical/object fields.
- `signup_date` is initially stored as an object/string and requires date parsing before date-related analysis.

## 4. Missing Values

Missing values are checked for every column to establish the completeness baseline.

```python
df.isnull().sum()
```

**Result:** No missing values were identified in the dataset.

## 5. Duplicate Records

A full-row duplicate check is performed to identify completely duplicated records.

```python
df.duplicated().sum()
```

**Result:** No full duplicate rows were identified.

## 6. Customer ID Uniqueness

The `customer_id` field is checked because it may serve as a candidate primary key.

```python
df["customer_id"].nunique()
```

**Result:** 100,000 unique customer IDs were found across 100,000 records.

This indicates that `customer_id` is unique within the dataset and is a candidate primary key.

## 7. Age Profiling

Basic descriptive statistics are used to understand the range and distribution of customer ages.

```python
df["age"].describe()
```

**Observed range:** ages are between 18 and 69.

This range was later used to define the validity rule for `age` in Stage 02.

## 8. Gender Distribution

```python
df["gender"].value_counts()
```

**Observed distribution:**
- Male: 50,120
- Female: 49,880

No unexpected gender category was identified during profiling.

## 9. Income Level Distribution

The frequency of each income category is reviewed to identify unexpected values or inconsistent representations.

```python
df["income_level"].value_counts()
```

**Initial observations:**
- High: 33,560
- LOW: 33,423
- Medium: 33,017

The value `LOW` was identified as a capitalization inconsistency compared with the expected standardized value `Low`. This observation was investigated further during the Data Quality Assessment.

## 10. Location Distribution

```python
df["location"].value_counts()
```

**Initial observations:**
- Mumbai: 16,859
- Bangalore: 16,669
- Chennal: 16,668
- Hyderabad: 16,634
- Kolkata: 16,603
- Delhi: 16,567

The value `Chennal` was identified as a naming inconsistency and was later standardized to `Chennai`.

## 11. Signup Date Profiling

The date field is converted from text to a datetime type so that its validity and range can be assessed.

```python
df["signup_date"] = pd.to_datetime(df["signup_date"])

df["signup_date"].min(), df["signup_date"].max()
```

**Observed date range:** 2017-01-01 to 2025-01-01.

The date values were successfully parsed and no invalid dates were identified during profiling.

# 12. Stage 01 Findings

The profiling established the following baseline:

| Area | Finding |
|---|---|
| Dataset size | 100,000 records × 6 columns |
| Missing values | None identified |
| Full duplicates | None identified |
| `customer_id` | 100,000 unique values; candidate primary key |
| `age` | Range: 18–69 |
| `gender` | Male and Female categories identified |
| `income_level` | `LOW` capitalization inconsistency observed |
| `location` | `Chennal` naming inconsistency observed |
| `signup_date` | Valid parsed dates from 2017-01-01 to 2025-01-01 |

### Key conclusion
Stage 01 established the baseline understanding of the dataset and identified a small number of observations requiring formal validation. These findings were carried forward into Stage 02 — Data Quality Assessment.
